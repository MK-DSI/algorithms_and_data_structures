```python
# Your answer here: 

# The task is to look at the binary tree ((using BFS (or Level Order Traversal) which
# traverses nodes present in the same level before traversing the next level))
# and determine if there are any duplicate numbers present in the tree. 
# If there are duplicate numbers (one type) - return that number.
# If there are multiple duplicates (different types) - only return a number closest to the root (beginning of the tree).
# If there are no duplicates - return [-1].
```

```python
# Your answer here
#My examples: 
    #Example1:  Input = [1,2,2,3,4,3,5]  (3 tiers, root  = 1, BFS, level order traversal)
    #           Expected Output = [2]

    #Example1:  Input = [2,4,6,6,8,8,10] (3 tiers, root  = 2, BFS, level order traversal)
    #           Expected Output = [6]

# Assignment partners' copied code: 
class TreeNode(object):   
    def __init__(self, val = 0, left = None, right = None):
        self.val = val  
        self.left = left
        self.right = right
        
def is_duplicate(root: TreeNode) -> int:
  # TODO

    # Store visited values and queue
    visited = set()
    queue = [root]

    while queue:
        # Pop the first node from the queue
        node = queue.pop(0)

        # Check if first node is visited
        if node.val in visited:
            # Return value if duplicated
            return node.val
        # Add current node to the visited set
        visited.add(node.val)

        # In case current node has a left child, add it to the queue
        if node.left:
            queue.append(node.left)
        # In case current node has a right child, add it to the queue
        if node.right:
            queue.append(node.right)
            
    # In case there is no duplicates in the tree, return -1
    return -1

# Test it
root = TreeNode(2)
root.left = TreeNode(4)
root.right = TreeNode(6)
root.left.left = TreeNode(6)
root.left.right = TreeNode(8)
root.right.left = TreeNode(8)
root.right.right = TreeNode(10)

print(is_duplicate(root))


# Explanation and walk through the code:

# First, a binary tree class "TreeNode" was created and value of the parent node 
# and left and right child nodes were defined. 

# Second, __init__ method initiates "TreeNode" class with values set to 
# val to 0, left node to none, right node to none.

# Third, is_duplicate funciton is defined:
        # Where 
             # A.  A new set "visited" is created
             # B.  A "queue" List is created with a root node. 


# Fourth, a 'while' loop function for "queue" is written to check 
# if there are nodes in the "queue". Then a "node" variable is assigned
# by removal of the first node from the "queue" - processing of the current node. 
# 
# Then 2 sections of the "while" function checking if the node.val was seen before 
# (by checking wiht the "visited" set)
# and if yes, then it is a duplicate adn the value is returned, if not, 
# it is added to the "visited" set.
#  
# Then checks for a left node is available, and if yes, then adds this left node to the queue.
# Similar, right node availability is checked and added to the queue if exists. 
# 
# WHile function ends
#  
# Fifth, if during while funciton no duplicates were found, the program returns [-1]

# Sixth, tree is constructued and checked. 

#Time and space complexity for this solution is O(n) 
```

```python
# Your answer here

# The solution provided is a high level which just indicated that instead of 
# BFS the traversal order needs to be DFS.
# In the DFS traversal order the traversion will go do the 
# depth exploring all levels first. My assignment partner was right in suggesting
# that the observed duplicates in this case may not be closest to the root. 
# 
# Potential solution to overcome this problem might be pre-order DFS traversal approach (extrenally found).
```

```python
def is_duplicate(root: TreeNode) -> int:
    # pre-order DFS traversal
    def podfs(node, visited):
        if not node:
            return -1
        
        # Check if the current node's value is in the visited set and adds
        if node.val in visited:
            return node.val
        visited.add(node.val)
        
        # Recursively check the left and righ subtrees
        left_result = podfs(node.left, visited)
        if left_result != -1:
            return left_result
        return podfs(node.right, visited)
    
    # Start DFS from the root
    visited = set()
    return podfs(root, visited)


# Testing
root = TreeNode(2)
root.left = TreeNode(4)
root.right = TreeNode(6)
root.left.left = TreeNode(6)
root.left.right = TreeNode(8)
root.right.left = TreeNode(8)
root.right.right = TreeNode(10)

print(is_duplicate(root))
```
